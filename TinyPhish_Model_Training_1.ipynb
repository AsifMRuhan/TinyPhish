```python

# STEP 3: Extract simple URL features

import re
import ipaddress
from urllib.parse import urlparse

def extract_features(url):
    url = str(url)
    parsed = urlparse(url if "://" in url else "//" + url)
    host = parsed.hostname or ""

    try:
        ipaddress.ip_address(host)
        has_ip = 1
    except ValueError:
        has_ip = 0

    return [
        len(url),                          # URL length
        url.count("."),                    # Number of dots
        url.count("-"),                    # Number of hyphens
        int(url.lower().startswith("https://")), # HTTPS flag
        int("@" in url),                   # Contains @
        sum(c.isdigit() for c in url),     # Number of digits
        url.count("/"),                    # Number of slashes
        int(has_ip)                        # Host is an IP address
    ]

# Test with one safe-looking and one suspicious-looking URL
test_urls = [
    "https://www.google.com",
    "http://192.168.1.1/login@verify"
]

for url in test_urls:
    print("\nURL:", url)
    print("Features:", extract_features(url))
```

```python

# STEP 11A: Compare Decision Tree configurations

from sklearn.model_selection import train_test_split
from sklearn.tree import DecisionTreeClassifier
from sklearn.metrics import classification_report
import pandas as pd

X_subtrain, X_val, y_subtrain, y_val = train_test_split(
    X_train, y_train,
    test_size=0.20,
    random_state=42,
    stratify=y_train
)

results = []

for depth in [5, 10, 15]:
    for leaf in [5, 20]:
        candidate = DecisionTreeClassifier(
            max_depth=depth,
            min_samples_leaf=leaf,
            class_weight="balanced",
            random_state=42
        )

        candidate.fit(X_subtrain, y_subtrain)
        pred = candidate.predict(X_val)

        report = classification_report(
            y_val, pred,
            output_dict=True,
            zero_division=0
        )

        results.append({
            "max_depth": depth,
            "min_samples_leaf": leaf,
            "phishing_precision": report["1"]["precision"],
            "phishing_recall": report["1"]["recall"],
            "phishing_f1": report["1"]["f1-score"],
            "accuracy": report["accuracy"]
        })

results_df = pd.DataFrame(results)
print(results_df.round(4).to_string(index=False))
```

```python

# STEP 12: Train selected model and export to JSON

from sklearn.tree import DecisionTreeClassifier
import json
import os

final_model = DecisionTreeClassifier(
    max_depth=15,
    min_samples_leaf=5,
    class_weight="balanced",
    random_state=42
)

# Train on the complete training set
final_model.fit(X_train, y_train)

# Export the learned tree
tree = final_model.tree_

tree_data = {
    "feature_names": feature_names,
    "children_left": tree.children_left.tolist(),
    "children_right": tree.children_right.tolist(),
    "features": tree.feature.tolist(),
    "thresholds": tree.threshold.tolist(),
    "values": tree.value[:, 0, :].tolist()
}

with open("phishing_tree.json", "w") as f:
    json.dump(tree_data, f, separators=(",", ":"))

size_bytes = os.path.getsize("phishing_tree.json")

print("Final tree depth:", final_model.get_depth())
print("Number of nodes:", final_model.tree_.node_count)
print("JSON model size:", round(size_bytes / 1024, 2), "KB")
print("Below 1 MB:", size_bytes < 1_000_000)
```

```python

# STEP 14: Test predictions from exported JSON

import json
import numpy as np

with open("phishing_tree.json", "r") as f:
    data = json.load(f)

def predict_json(row):
    node = 0

    while data["children_left"][node] != -1:
        feature = data["features"][node]
        threshold = data["thresholds"][node]

        if row[feature] <= threshold:
            node = data["children_left"][node]
        else:
            node = data["children_right"][node]

    return int(np.argmax(data["values"][node]))

# Test the first 100 URLs
X_sample = X_test.iloc[:100].to_numpy()

json_predictions = np.array([
    predict_json(row) for row in X_sample
])

python_predictions = final_model.predict(X_test.iloc[:100])

print("Predictions match:",
      np.array_equal(json_predictions, python_predictions))

print("Python model:", python_predictions[:10].tolist())
print("JSON model:  ", json_predictions[:10].tolist())
```

```python
# STEP 6: Train an experimental model on PhishBD_2026
# This does NOT replace the current Android model.

import numpy as np
from sklearn.model_selection import train_test_split
from sklearn.tree import DecisionTreeClassifier
from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    confusion_matrix,
    classification_report
)

# Keep the same eight features used by TinyPhish.
# Convert at_count into the binary has_at feature.
X = pd.DataFrame({
    "url_length": phishbd_df["url_length"],
    "num_dots": phishbd_df["dot_count"],
    "num_hyphens": phishbd_df["hyphen_count"],
    "has_https": phishbd_df["is_https"],
    "has_at": (phishbd_df["at_count"] > 0).astype(int),
    "num_digits": phishbd_df["digit_count"],
    "num_slashes": phishbd_df["slash_count"],
    "has_ip": phishbd_df["has_ip"]
})

y = phishbd_df["label"].astype(int)

# Reserve 20% of the data for testing.
X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42,
    stratify=y
)

# Train a small decision tree, matching the current depth limit.
model_bd_baseline = DecisionTreeClassifier(
    max_depth=10,
    class_weight="balanced",
    random_state=42
)

model_bd_baseline.fit(X_train, y_train)

# Predict on the held-out test set.
y_pred = model_bd_baseline.predict(X_test)

print("Training rows:", len(X_train))
print("Testing rows:", len(X_test))
print("Model depth:", model_bd_baseline.get_depth())
print("Model leaves:", model_bd_baseline.get_n_leaves())

print("\n--- Evaluation results ---")
print("Accuracy:", round(accuracy_score(y_test, y_pred), 4))
print("Precision:", round(precision_score(y_test, y_pred, zero_division=0), 4))
print("Recall:", round(recall_score(y_test, y_pred, zero_division=0), 4))
print("F1-score:", round(f1_score(y_test, y_pred, zero_division=0), 4))

print("\nConfusion matrix:")
print(confusion_matrix(y_test, y_pred, labels=[0, 1]))

print("\nClassification report:")
print(classification_report(
    y_test,
    y_pred,
    labels=[0, 1],
    target_names=["Legitimate (0)", "Phishing (1)"],
    zero_division=0
))
```

```python

# STEP 26.2: Train and evaluate three candidate models
# This does not replace or export the final Android model.

import pandas as pd

from sklearn.pipeline import make_pipeline
from sklearn.preprocessing import StandardScaler
from sklearn.linear_model import LogisticRegression
from sklearn.tree import DecisionTreeClassifier
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    confusion_matrix
)

# 1. Define the candidate models.
comparison_models = {
    "Logistic Regression": make_pipeline(
        StandardScaler(),
        LogisticRegression(
            class_weight="balanced",
            max_iter=1000,
            random_state=42
        )
    ),

    "Decision Tree": DecisionTreeClassifier(
        max_depth=10,
        class_weight="balanced",
        random_state=42
    ),

    "Random Forest": RandomForestClassifier(
        n_estimators=100,
        max_depth=10,
        class_weight="balanced",
        random_state=42,
        n_jobs=-1
    )
}

# 2. Train and evaluate each model on the same split.
comparison_results = []
comparison_confusion_matrices = {}

for model_name, model in comparison_models.items():
    print(f"Training {model_name}...")

    model.fit(X_train_compare, y_train_compare)

    # Probability of class 1.
    # This assumes label 1 means phishing.
    phishing_class_index = list(model.classes_).index(1)
    phishing_probabilities = model.predict_proba(
        X_test_compare
    )[:, phishing_class_index]

    # Use the same 0.30 threshold for all models.
    predictions = (
        phishing_probabilities >= 0.30
    ).astype(int)

    comparison_results.append({
        "Model": model_name,
        "Accuracy": accuracy_score(
            y_test_compare, predictions
        ),
        "Precision": precision_score(
            y_test_compare, predictions,
            zero_division=0
        ),
        "Recall": recall_score(
            y_test_compare, predictions,
            zero_division=0
        ),
        "F1-score": f1_score(
            y_test_compare, predictions,
            zero_division=0
        )
    })

    comparison_confusion_matrices[model_name] = (
        confusion_matrix(
            y_test_compare,
            predictions,
            labels=[0, 1]
        )
    )

# 3. Display the comparison.
comparison_table = pd.DataFrame(comparison_results)

metric_columns = [
    "Accuracy", "Precision", "Recall", "F1-score"
]

comparison_table[metric_columns] = (
    comparison_table[metric_columns].round(4)
)

print("\nMODEL COMPARISON (threshold = 0.30)")
display(comparison_table)

# 4. Show confusion matrices.
for model_name, matrix in comparison_confusion_matrices.items():
    print(f"\n{model_name} confusion matrix")
    print("Rows = actual [legitimate, phishing]")
    print("Columns = predicted [legitimate, phishing]")
    print(matrix)
```

```python

# STEP 40: Verify exported JSON tree predictions

import numpy as np

def predict_from_exported_tree(row, data, threshold=0.30):
    node = 0

    while data["children_left"][node] != -1:
        feature_index = data["features"][node]
        split_threshold = data["thresholds"][node]

        if row[feature_index] <= split_threshold:
            node = data["children_left"][node]
        else:
            node = data["children_right"][node]

    benign, phishing = data["values"][node]
    total = benign + phishing

    probability = phishing / total if total > 0 else 0.0
    return int(probability >= threshold), probability

# Use the same test rows and threshold as before.
test_rows = X_test.iloc[:20].to_numpy(dtype=float)
sklearn_probabilities = model_bd_baseline.predict_proba(
    X_test.iloc[:20]
)[:, 1]

export_predictions = []
export_probabilities = []

for row in test_rows:
    prediction, probability = predict_from_exported_tree(
        row, experimental_model_data, threshold=0.30
    )
    export_predictions.append(prediction)
    export_probabilities.append(probability)

print("Scikit-learn probabilities:")
print(np.round(sklearn_probabilities, 6))

print("\nExported JSON probabilities:")
print(np.round(export_probabilities, 6))

print("\nPredictions match:",
      np.array_equal(
          (sklearn_probabilities >= 0.30).astype(int),
          np.array(export_predictions)
      ))

print("Maximum probability difference:",
      round(float(np.max(np.abs(
          sklearn_probabilities - np.array(export_probabilities)
      ))), 10))
```
